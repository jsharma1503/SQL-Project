```python
import sqlite3
```

```python
import pandas as pd
```

# Database Creation

```python
# Create Database for Online Book Store
conn = sqlite3.connect("Online Book Store.db")
cursor = conn.cursor()
```

```python
# Creating Table for Books Information in Database
cursor.execute("""
    DROP TABLE IF EXISTS Books
""")
conn.commit()

cursor.execute("""
    CREATE TABLE Books(
    Book_ID INTEGER PRIMARY KEY,
    Title VARCHAR(100),
    Author VARCHAR(100),
    Genre VARCHAR(50),
    Published_Year INTEGER,
    Price NUMERIC(10, 2),
    Stock INTEGER
    );
""")
conn.commit()
```

```python
# Adding Books Information in Database
df = pd.read_csv('Books.csv')
df.to_sql(
    "Books",      
    conn,
    if_exists="replace",
    index=False
)
```

```python
# Creating Table for Customers Information in Database
cursor.execute("""
    DROP TABLE IF EXISTS Customers
""")
conn.commit()

cursor.execute("""
    CREATE TABLE Customers(
    Customer_ID INTEGER PRIMARY KEY,
    Name VARCHAR(100),
    Email VARCHAR(100),
    Phone INTEGER,
    City VARCHAR(50),
    Country VARCHAR(50)
    );
""")
conn.commit()
```

```python
# Adding Customers Information in Database
df = pd.read_csv('Customers.csv')
df.to_sql(
    "Customers",
    conn,
    if_exists="replace",
    index=False
)
```

```python
# Creating Table for Orders Information in Database
cursor.execute("""
    DROP TABLE IF EXISTS Orders
""")
conn.commit()

cursor.execute("""
    CREATE TABLE Orders(
    Order_ID INTEGER PRIMARY KEY,
    Book_ID INTEGER,
    Customer_ID INTEGER,
    Order_Date DATE,
    Quantity INTEGER,
    Total_Amount NUMERIC(10, 2)
    );
""")
conn.commit()
```

```python
# Adding Orders Information in Database
df = pd.read_csv('Orders.csv')
df.to_sql(
    "Orders",
    conn,
    if_exists="replace",
    index=False
)
```
